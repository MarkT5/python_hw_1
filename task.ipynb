```python
def print_mid(txt):
  mid = len(txt)//2
  if len(txt) == 0:
    print("Empty string")
  elif len(txt) % 2:
    print(txt[mid])
  else:
    print(txt[mid-1:mid+1])
```

```python
print_mid(input())
```

```python
print_mid(input())
```

```python
def print_date(male, female):
  if len(male) != len(female):
    print("Внимание, кто-то может остаться без пары!")
    return
  male.sort()
  female.sort()
  print("Идеальные пары:")
  for m, f in zip(male, female):
    print(m, "и", f)
```

```python
boys = ['Peter', 'Alex', 'John', 'Arthur', 'Richard']
girls = ['Kate', 'Liza', 'Kira', 'Emma', 'Trisha']
```

```python
print_date(boys, girls)
```
